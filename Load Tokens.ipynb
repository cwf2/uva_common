# Load Tokens

Minimal example: download the pre-built token tables for the corpus and load them into a single pandas DataFrame. See `Build Token Tables.ipynb` for how these tables are generated, and the README for background on the corpus.

```python
# install common code for UvA Digital Classics
%pip install git+https://github.com/cwf2/uva_common

# import statements
import os
import pandas as pd
import uva_common
```

```python
# select text(s) to analyze
#

urns = [
    "tlg0012.tlg001.perseus-grc2", # Iliad
    "tlg0012.tlg002.perseus-grc2", # Odyssey
    "tlg0001.tlg001.perseus-grc2", # Argonautica
    "tlg2046.tlg001.perseus-grc2", # Posthomerica
    "tlg0647.tlg001.perseus-grc2", # Sack of Troy
    "tlg2045.tlg001.perseus-grc2", # Dionysiaca
]

# actual filenames have csv extension
filenames = [urn + ".csv" for urn in urns]

# download the token data from OSF
uva_common.download(filenames, node_id="tokens")
```

```python
# read all the downloaded files into one dataframe
tokens = pd.concat(pd.read_csv(os.path.join("data", filename)) for filename in filenames)
display(tokens)
```
